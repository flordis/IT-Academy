### 1. Calculadora de l'índex de massa corporal

- Escriu una funció que calculi l'IMC **ingressat per l'usuari/ària**, és a dir, qui ho executi haurà d'ingressar aquestes dades.
- La funció ha de **classificar el resultat** en les seves respectives categories.

Consell: Intenta **validar les dades prèviament**, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.

```python
def BMI_calculation(decimals=2):
    while True:
        try:
            weight = float(input("Please, enter your weight in kilograms (15-300)."))
            if 15 <= weight <= 300:
                print(f"Weight entered successfully: {weight} kg.")
                break    
            else:
                print("Invalid input. Weight must be between 15 and 300 kilograms. Please, try again.")    
        except ValueError:
            print("Invalid format. Please, make sure to enter ONLY a positive number (e.g., 70.5).")
        except TypeError:
            print("Invalid format. Please, make sure to enter ONLY a positive number (e.g., 70.5).")
        
    while True:
        try:
            height = float(input("Please, enter your height in meters (1.0-2.5)."))
            if 1.0 <= height <= 2.5:
                print(f"Height entered successfully: {height} m.")
                break
            else:
                print("Invalid input. Height must be between 1.0 and 2.5 meters. Please, try again.")
        except ValueError:
            print("Invalid format. Please, make sure to enter ONLY a positive number (e.g., 1.75).")
        except TypeError:
            print("Invalid format. Please, make sure to enter ONLY a positive number (e.g., 1.75).")

    bmi = round(weight / (height ** 2), decimals)
    
    if bmi < 18.5:
        print(f"Your BMI is: {bmi}. You are underweight.")
    elif bmi <= 24.9:
        print(f"Your BMI is: {bmi}. You are within the optimum range.")
    elif bmi <= 29.0:
        print(f"Your BMI is: {bmi}. You are overweight.")
    else:
        print(f"Your BMI is: {bmi}. You are within the obese range.")
    return bmi

BMI_calculation()
```

### 2. Convertidor de temperatures

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).  
- Selecciona **almenys 2 conversors**, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera **que es puguin guardar** (recorda que un print() no es pot guardar mai).

Consell: Intenta **validar les dades prèviament**, perquè enviï un missatge d’advertència si les dades introduides per l’usuari/a no estan en el format adequat.

- (EXTRA): Pensa una **manera d’emmagatzemar totes les possibles conversions en un sol objecte** (Llista? Diccionari? DataFrame?) en comptes d’escriure molts if else en funció de la temperatura d’origen i la temperatura de destí.

```python
# CORRECTION
def temperature_conversor_extra():
    units_info = {                      #  main dictionary
        "C": {
            "name": "Celsius",
            "symbol": "ºC",
            "min_value": -273.15,
            "max_value": 1000,
            "factor": 1,
            "offset": 273.15
        },
        "F": {
            "name": "Fahrenheit",
            "symbol": "ºF",
            "min_value": -459.67,
            "max_value": 1832,
            "factor": 5/9,
            "offset": -32
        },
        "K": {
            "name": "Kelvin",
            "symbol": "K",
            "min_value": 0,
            "max_value": 1273.15,
            "factor": 1,
            "offset": 0
        }
    }
    
    get_aliases = lambda unit_code: [unit_code, units_info[unit_code]["symbol"], units_info[unit_code]["name"]]
    
    all_valid_units = []
    for code in units_info.keys():
        all_valid_units.extend(get_aliases(code))
    
    validate_range = lambda unit_code, value: units_info[unit_code]["min_value"] <= value <= units_info[unit_code]["max_value"]
    
    while True:
        origin_input = str(input("Please, enter the origin unit (C, ºC, Celsius, K, ºK, Kelvin, F, ºF, Fahrenheit): ")).capitalize()
        if origin_input in all_valid_units:
            origin = next(code for code in units_info.keys() if origin_input in get_aliases(code))
            print(f"Origin unit entered successfully: {origin_input}.")
            break
        else:
            print("Invalid input. Please, enter a valid unit (C, ºC, Celsius, K, Kelvin, F, ºF, Fahrenheit).")
    
    while True:
        try:
            value = float(input("Please, enter the temperature value to be converted: "))
            if validate_range(origin, value):
                print(f"Temperature value entered successfully: {value} {units_info[origin]['symbol']}.")
                break
            else:
                print(f"Invalid range. Please enter a value between {units_info[origin]['min_value']} and {units_info[origin]['max_value']} for {units_info[origin]['name']}.")
        except ValueError:
            print("Invalid format. Please, make sure to enter a valid NUMBER.")

    kelvin = (value + units_info[origin]["offset"]) * units_info[origin]["factor"]
    
    conversions = {}
    
    for target_code, target_info in units_info.items():
        if target_code == origin:
            continue
        converted = kelvin / target_info["factor"] - target_info["offset"]
        conversions[target_code] = converted
        print(f"{value}{units_info[origin]['symbol']} = {round(converted, 2)}{target_info['symbol']} ({target_info['name']})")
    
    return conversions

temperature_conversor_extra()
```

### 3. Comptador de paraules d'un text.

- Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

(EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3

```python
# CORRECTION
def text_input():    
    while True:
        user_text = str(input("Please, enter the text: "))
        if not user_text or user_text.isspace():
            print("Error: Empty text or only spaces detected. Please enter at least one word.")
        else:
            break
    return user_text
```

```python
def word_count(validated_text):
    """
    Counts word frequencies in a text, orders words alphabetically,
    and returns a dictionary with word:count.
    """
    text_list = re.findall(r'\b\w+\b', validated_text.lower())
    text_list.sort()  
    word_count_file = dict(Counter(text_list))
    return word_count_file
```

```python
def word_length_avg(word_dict):
    """
    Calculates the average word length from a dictionary of words.
    Uses mean() and map() in a single line as suggested.
    """
    return mean(map(len, word_dict))
```

```python
def main_function():
    validated_text = text_input()
    print(f"\nText validated: \"{validated_text}\"\n")
    word_frequencies = word_count(validated_text)
    print(f"Word frequencies calculated: {word_frequencies}\n")
    average_length = word_length_avg(word_frequencies)
    print(f"\nAverage word length: {average_length:.2f}") 
    return {
        "original_text": validated_text,
        "word_frequencies": word_frequencies,
        "average_word_length": average_length
    }

main_function()
```

### 4. Diccionari invers (amb possibilitat de duplicats)

- Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

```python
# CORRECTION
def reverse_dictionary(dict):
    inverted_dict = {}
    duplicates = []
    for key, value in dict.items():
        if value not in inverted_dict:
            inverted_dict[value] = key
        else:
            if not isinstance(inverted_dict[value], list):
                inverted_dict[value] = [inverted_dict[value]]
            inverted_dict[value].append(key)
            if value not in duplicates:
                duplicates.append(value)
    if duplicates:
        print(f"\nWarning: The following values have more than one key: {duplicates}")
    return inverted_dict

test_dictionary1 = {'a': 1, 'b': 2, 'c': 3}
test_1 = reverse_dictionary(test_dictionary1)
print(test_1)

test_dictionary2 = {'x': 'apple', 'y': 'banana', 'z': 'banana'}
test_2 = reverse_dictionary(test_dictionary2)
print(test_2)
```

Per exemple, per a l'arxiu "tu_me_quieres_blanca.txt" la sortida esperada seria:

```python
def file_input(file_path):    
    file_to_process = file_path
    try:
        with open(file_to_process, encoding='utf-8') as raw_file:
            content = raw_file.read()
        return content
    except FileNotFoundError:
        print("Error: The file does not exist.")
        return None
    except PermissionError:
        print("Error: You do not have permission to access this file.")
        return None
    except IsADirectoryError:
        print("Error: You have entered a directory instead of a file. Please, enter a valid file path.")
        return None
    except UnicodeDecodeError:
        print("Error: The file contains characters that cannot be decoded with UTF-8 encoding.")
        return None

validated_user_file = file_input(input("Please, enter the path of the txt file to process."))
print(validated_user_file)
```

```python
def word_count_order(validated_text):
    include = r'\b[a-zA-ZáéíóúüñÁÉÍÓÚÜÑ]+\b'
    text_list = re.findall(include, validated_text.lower())
    ordered_text_list = sorted(text_list)
    
    word_count_file = {}
    for word in ordered_text_list:
        word_lower = word.lower()
        if word_lower in word_count_file:
            word_count_file[word_lower] += 1
        else:
            word_count_file[word_lower] = 1

    nested_dictionary = {}
    for word, count in word_count_file.items():
        initial = word[0].lower()
        if initial not in nested_dictionary.keys():
            nested_dictionary[initial] = {word: count}
    
    return nested_dictionary

file = word_count_order(validated_user_file)
pprint.pprint(file)
```

### 2. Conversió de tipus de dades.

- El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir.

```python
def data_type_converter(list_to_process):
    float_elements_list = []
    non_float_elements_list = []

    for element in list_to_process:
        if isinstance(element, (list, tuple)):                      
            floats, non_floats = data_type_converter(element)
            float_elements_list.extend(floats)
            non_float_elements_list.extend(non_floats)
        else:
            try:
                if isinstance(element, str): 
                    element = element.replace(',', '.')
                float_value = float(element)
                float_elements_list.append(float_value)
            except (ValueError, TypeError, Exception):
                non_float_elements_list.append(element)

    return float_elements_list, non_float_elements_list

client_list = ['1.3', 'one', '1e10', 'seven', '3-1/2', ('2', '1', '1,4', 'not-a-number'), [1, 2, '3', '3.4']]
result = data_type_converter(client_list)
print(result)
```

### 1. Generador de Contrasenyes

- Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que **generi contrasenyes més segures**. La funció ha de dependre dels següents paràmetres:

    longitud (int): Longitud de la contrasenya
    majuscules (bool = True): Si hi ha d’aparèixer majúscules
    minuscules (bool = True): Si hi ha d’aparèixer minúscules
    numeros (bool = True): Si hi ha d’aparèixer números
    signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)

- Així doncs, si executem la funció de la següent manera:

crear_contrasenya(10, True, True, True, True)

- Hauríem d’obtenir un output (que **hem de poder guardar**) de l’estil:

9Er,5Vn8P$

- Assegura’t que es compleixin **tots els criteris**, i que aquestes contrasenyes siguin **realment aleatòries**.

- (EXTRA) Explora com podríem fer que la **funció copiés la contrasenya automàticament al porta papers de l’ordinador** (com si l’haguéssim seleccionat i fet ctrl+copy).

```python
def key_generator(length, uppercase, lowercase, numbers, special_characters):  
    character_types = []
    all_characters = ''
    
    if uppercase:
        character_types.append(string.ascii_uppercase)
        all_characters += string.ascii_uppercase
    
    if lowercase:
        character_types.append(string.ascii_lowercase)
        all_characters += string.ascii_lowercase
    
    if numbers:
        character_types.append(string.digits)
        all_characters += string.digits
    
    if special_characters:
        character_types.append(string.punctuation)
        all_characters += string.punctuation

    if not all_characters:
        raise ValueError("At least one character type must be selected.")
    
    selected_characters = []

    for each_type in character_types:
        character = np.random.choice(list(each_type))
        selected_characters.append(character)
        
    remaining = length - len(selected_characters)
    if remaining > 0:
        extra_characters = np.random.choice(list(all_characters), size=remaining)
        selected_characters.extend(extra_characters)
    
    random.shuffle(selected_characters)
    key = ''.join(selected_characters)
    
    pyperclip.copy(key)
    return key

random_key = key_generator(10, True, True, True, True)
print(random_key)
```

### 2. Processament de dades simple

- Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.

- Utilitza el fitxer "historic_partits.txt"

- Necessita un programa que retorni:

--    El nombre total de gols que ha fet cada equip.
--    El nom de l’equip més golejador.
--    El nom de l’equip més golejat
--    La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)

- Com a referència, els equips amb més punts serien els següents: (veure online)

```python
def process_match_results(file_path):
    df = pd.read_csv(file_path, sep='\t|-', header=None, engine='python')

    df.columns = ['home_team', 'goals_home', 'goals_visiting', 'visiting_team']
    df.astype({'goals_home': int, 'goals_visiting': int})
    df = df.iloc[:, [0, 1, 3, 2]]
    
    df['points_home'] = ((df['goals_home'] > df['goals_visiting']) * 3 + (df['goals_home'] == df['goals_visiting']) * 1)
    df['points_visiting'] = ((df['goals_visiting'] > df['goals_home']) * 3 + (df['goals_visiting'] == df['goals_home']) * 1)

    teams = pd.concat([df['home_team'], df['visiting_team']]).unique()

    win_occurrences_home = df[df['points_home'] == 3].groupby('home_team').size()
    win_occurrences_visiting = df[df['points_visiting'] == 3].groupby('visiting_team').size()
    wins = win_occurrences_home.add(win_occurrences_visiting)

    draw_occurrences_home = df[df['points_home'] == 1].groupby('home_team').size()
    draw_occurrences_visiting = df[df['points_visiting'] == 1].groupby('visiting_team').size()
    draws = draw_occurrences_home.add(draw_occurrences_visiting)

    loss_occurrences_home = df[df['points_home'] == 0].groupby('home_team').size()
    loss_occurrences_visiting = df[df['points_visiting'] == 0].groupby('visiting_team').size()
    losses = loss_occurrences_home.add(loss_occurrences_visiting)

    for_home = df.groupby('home_team')['goals_home'].sum()
    for_visiting = df.groupby('visiting_team')['goals_visiting'].sum()
    goals_for = for_home.add(for_visiting)

    against_home = df.groupby('home_team')['goals_visiting'].sum()
    against_visiting = df.groupby('visiting_team')['goals_home'].sum()
    goals_against = against_home.add(against_visiting)

    point_home = df.groupby('home_team')['points_home'].sum()
    point_visiting = df.groupby('visiting_team')['points_visiting'].sum()
    points = point_home.add(point_visiting)

    new_df = pd.DataFrame({
        'wins': [wins.get(t, 0) for t in teams],
        'draws': [draws.get(t, 0) for t in teams],
        'losses': [losses.get(t, 0) for t in teams],
        'goals_for': [goals_for.get(t, 0) for t in teams],
        'goals_against': [goals_against.get(t, 0) for t in teams],
        'points': [points.get(t, 0) for t in teams]
    }, index=teams)
    
    new_df = new_df.sort_values(by='points', ascending=False)
    return new_df

file_to_process = r'C:\Users\Asus\Documents\IT\Especialidad_IT_Academy\Sprint 9\historic_partits.txt'
result = process_match_results(file_to_process)

max_team = result['goals_for'].idxmax()
max_goals = result.loc[max_team, 'goals_for']
    
min_team = result['goals_for'].idxmin()
min_goals = result.loc[min_team, 'goals_for']

print(f"The team with more goals is: {max_team} with {max_goals} goals")
print(f"The team with fewer goals is: {min_team} with {min_goals} goals")
print(result)
```
