```python
def BMI_calculation(decimals = 2):

# This funtion asks for weight and height, and returns the BMI value and the corresponding category (underweight, optimum range, overweight, obese)
# Default decimals = 2, but can be changed by user when calling the function

    while True:
        try:
            weight = float(input("Please, enter your weight in kilograms (15-300)."))
            if 15 <= weight <=300:
                print(f"Weight entered successfully: {weight} kg.")
                break    
            else:
                print("Invalid input. Weight must be between 15 and 300 kilograms. Please, try again.")    
        except ValueError:            
            print("Invalid format. Please, make sure to enter ONLY a positive number (e.g., 70.5).")
        except TypeError:
            print("Invalid format. Please, make sure to enter ONLY a positive number (e.g., 70.5).")
        
    while True:
        try:
            height = float(input("Please, enter your height in meters (1.0-2.5)."))
            if 1.0 <= height <=2.5:
                print(f"Height entered successfully: {height} m.")
                break
            else:
                print("Invalid input. Height must be between 1.0 and 2.5 meters. Please, try again.")
        except ValueError:
            print("Invalid format. Please, make sure to enter ONLY a positive number (e.g., 1.75).")
        except TypeError:
            print("Invalid format. Please, make sure to enter ONLY a positive number (e.g., 1.75).")

    bmi = round(weight / (height **2), decimals)
    
    if bmi < 18.5:
        print(f"Your BMI is: {bmi}. You are underweight.")
    elif bmi <= 24.9:
        print(f"Your BMI is: {bmi}. You are within the optimum range.")
    elif bmi <= 29.0:
        print(f"Your BMI is: {bmi}. You are overweight.")
    else:
        print(f"Your BMI is: {bmi}. You are within the obese range.")
    return bmi  # Not asked for in the instructons, but I think it is useful to return the BMI value for further use, is needed.

BMI_calculation()
```

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).  
- Selecciona **almenys 2 conversors**, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera **que es puguin guardar** (recorda que un print() no es pot guardar mai).

```python
# Temperature conversion function 

def temperature_conversor():
    # Validation of user input for origin
    while True:
        origin = str(input("Please, enter the origin unit (C, ºC, Celsius, K, ºK, Kelvin, F, ºF, Fahrenheit).")).capitalize()
        if origin in ["C", "ºC", "Celsius", "K", "ºK", "Kelvin", "F", "ºF", "Fahrenheit"]:
            print(f"Origin unit entered successfully: {origin}.")
            break
        else:
            print("Invalid input. Please, enter a valid unit (C, ºC, Celsius, K, Kelvin, F, ºF, Fahrenheit).")
    
    # Validation of user input for destination
    while True:
        destination = str(input("Please, enter the destination unit (C, ºC, Celsius, K, ºK, Kelvin, F, ºF, Fahrenheit).")).capitalize()
        if destination in ["C", "ºC", "Celsius", "K", "ºK", "Kelvin", "F", "ºF", "Fahrenheit"]:
            print(f"Destination unit entered successfully: {destination}.")
            break
        else:
            print("Invalid input. Please, enter a valid unit (C, ºC, Celsius, K, Kelvin, F, ºF, Fahrenheit).")

    # Validation of user input for value
    while True:
        try:
            value = float(input("Please, enter the temparature value to be converted (between -273.15 and 1000 for Celsius, 0 and 1273.15 for Kelvin, -459.67 and 1832 for Fahrenheit)."))
            if origin in ["C", "ºC", "Celsius"] and -273.15 <= value <= 1000:
                print(f"Temperature value entered successfully: {value} ºC.")
                break
            elif origin in ["K", "ºK", "Kelvin"] and 0 <= value <= 1273.15:
                print(f"Temperature value entered successfully: {value} K.")
                break
            elif origin in ["F", "ºF", "Fahrenheit"] and -459.67 <= value <= 1832:
                print(f"Temperature value entered successfully: {value} ºF.")
                break
        except ValueError:
            print("Invalid format. Please, make sure to enter a value within the valid range for the selected unit.")
        except TypeError:
            print("Invalid format. Please, make sure to enter a valid NUMBER within the valid range for the selected unit.")
    
    # Normalization to Kelvin
    if origin == "C":
        kelvin = value + 273.15
    elif origin == "F":
        kelvin = (value - 32) * 5/9 + 273.15
    elif origin == "K":
        kelvin = value

    # Conversion from Kelvin to destination unit
    converted_value = None
    if destination == "C":
        converted_value = kelvin - 273.15
    elif destination == "F":
        converted_value = (kelvin - 273.15) * 9/5 + 32
    elif destination == "K":
        converted_value = kelvin
    
    print(f"\n{value}°{origin} = {round(converted_value, 2)}°{destination}")
    
    return round(converted_value, 2)

temperature_conversor()
```

```python
# This function asks user to enter a text and validates it
# Possible errors: empty text or just spaces

def text_input():    
    while True:
        user_text = str(input("Please, enter the text."))

        if not user_text:
            print("Error: You entered an empty text. Please enter at least one word.")
        elif len(user_text) == 0:
            print("Error: Only spaces detected. Please enter actual words.")
        else:
            break
    
    return user_text

validated_user_text = text_input()

print(validated_user_text)      # Just to show the validated text, but it is not required by the instructions.
```

```python
# This function counts the number of times each word repeats in a given text, orders the words alphabetically and returns a dictionary with that information.

def word_count(validated_text):
        
    # user_text into a clean list of words
    text_list = re.findall(r'\b\w+\b', validated_text.lower())        
    
    # order words alphabetically (not requied, but it is a good practice to have the output ordered)
    text_list.sort()

    # Empty dictionary to populate it with words (key) and count (value)
    word_count_file = {}
    
    # Loop to populate the dictionary
    for word in text_list:
        if word in word_count_file:
            word_count_file[word] += 1    # Word already exists, increment count
        else:
            word_count_file[word] = 1     # New word, add to dictionary with count 1

    return word_count_file

# Function call and storage
# In this case, the parameter is the text from the previous function

file = word_count(validated_user_text)

print(file)     # Just to show the output, but it is not required by the instructions.
```

### 4. Diccionari invers (amb possibilitat de duplicats)

- Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

```python
# This function receives a dictionary and returns a new dictionary with keys and values inverted. If there are duplicate values, the new value will be a list of keys.

def reverse_dictionary(dict):
    
    # Empty dictionary and empty list for duplicates
    inverted_dict = {}
    duplicates = []
    
    for key, value in dict.items():
        if value not in inverted_dict:
            inverted_dict[value] = [key]                # If it doesn't exist, a new key is created (with the value)
        else:                                           # [key] is a list, in order to be able to use append later
            inverted_dict[value].append(key)            # If it exists, a key is added to the list (with the value)
            if value not in duplicates:
                duplicates.append(value)
        
    if duplicates:
        print(f"\nWarning: The following keys have more than one value: {duplicates}") 
    
    return inverted_dict

test_dictionary1 = {'a':1, 'b':2, 'c':3}

test_1 = reverse_dictionary(test_dictionary1)

print(test_1)

test_dictionary2 = {'x': 'apple', 'y': 'banana', 'z':'banana'}

test_2 = reverse_dictionary(test_dictionary2)

print(test_2)
```

```python
# This function receives a txt path, opens the file, reads it and validates it
# Possible errors: FileNotFound, PermissionError, IsADirectoryError, UnicodeDecodeError

def file_input(file_path):    
    
    file_to_process = file_path
    try:
        with open(file_to_process, encoding='utf-8') as raw_file:
            content = raw_file.read()
        return content
    
    # Error validation: file not found
    except FileNotFoundError:
        print("Error: The file does not exsist.")
        return None
    
    # Error validation: permission error
    except PermissionError:
        print("Error: You do not have permission to access this file.")
        return None
    
    # Error validation: is a directory
    except IsADirectoryError:
        print("Error: You have entered a directory instead of a file. Please, enter a valid file path.")
        return None
    
    #Error validation: Unicode decode error
    except UnicodeDecodeError:
        print("Error: The file contains characters that cannot be decoded with UTF-8 encoding. Please, check the file encoding or try with another file.")
        return None

validated_user_file = file_input(input("Plase, enter the path of the txt file to process."))

print(validated_user_file)
```

```python
# This function counts the number of times each word repeats in a given text, orders the words alphabetically and returns a dictionary with that information.
# The difference with the previous one is that this one receives a text as a parameter (from a path, in this case), instead of using the variable "file", which comes from a user input.
# It can be used with any text, not only with the one from the previous function.

def word_count_order(validated_text):
        
    # user_text into a clean list of words (characters and lower case)
    include = r'\b[a-zA-ZáéíóúüñÁÉÍÓÚÜÑ]+\b'
    text_list = re.findall(include, validated_text.lower())
    
    # order words alphabetically
    ordered_text_list = sorted(text_list)
    
    # Empty dictionary to populate it with {word (key) : count (value)}
    word_count_file = {}
    
    # Loop to populate the dictionary with {word(key):count(value)}
    for word in ordered_text_list:
       word_lower = word.lower()                # Convert to lowercase to count "Hello" and "hello" as the same word
       if word_lower in word_count_file:
            word_count_file[word_lower] += 1    # Word already exists, increment count
       else:
            word_count_file[word_lower] = 1       # New word, add to dictionary with count 1

    # Empty dictionary to populate it with {{initial(key) : { word (key) : count (value) } / {{initial : {word_count_file}
    nested_dictionary = {}
    
    # Loop to set first letter as key and populate the dictionary
    for word, count in word_count_file.items():
        initial = word[0]                      
        if initial not in nested_dictionary.keys():
           nested_dictionary[initial] = {word:count}        # If initial does not exist in dictionary, a new key is created 
        else:
           nested_dictionary[initial][word] = count          # If initial exists in dictionary, the word and count are added to the nested dictionary

    return nested_dictionary

# Function call and storage
# In this case, the parameter is the text from the previous function

file = word_count_order(validated_user_file)

pprint.pprint(file)
```

### 2. Conversió de tipus de dades.

- El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir.

```python
# This function receives a list of elements and separates the float elements from the non-float elements, 
# ...returning two lists: one with the float elements and another with the non-float elements.

def data_type_converter(list_to_process):
    
    # Empty lists to populate with float and non-float elements
    float_elements_list = []
    non_float_elements_list = []

    # Loop to separate float and non-float elements
    for element in list_to_process:
        
        # if the element is a list or tuple, the same function we're creating is used to process it (recursive call):
        if isinstance(element, (list, tuple)):                      
            floats, non_floats = data_type_converter(element)       # RECURSIVE CALL
            float_elements_list.extend(floats)                      
            non_float_elements_list.extend(non_floats)
        
        # if the element is a string
        else:
            try:  
                if isinstance(element, str): 
                    element = element.replace(',', '.')
                float_value = float(element)
                float_elements_list.append(float_value)
            except (ValueError, TypeError, Exception):
                non_float_elements_list.append(element)

    return float_elements_list, non_float_elements_list

client_list = ['1.3', 'one', '1e10', 'seven', '3-1/2', ('2', '1', '1,4', 'not-a-number'), [1, 2, '3', '3.4']]

result = data_type_converter(client_list)
print(result)
```

### 1. Generador de Contrasenyes

- Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

    longitud (int): Longitud de la contrasenya
    majuscules (bool = True): Si hi ha d’aparèixer majúscules
    minuscules (bool = True): Si hi ha d’aparèixer minúscules
    numeros (bool = True): Si hi ha d’aparèixer números
    signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)

- Així doncs, si executem la funció de la següent manera:

- Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:

9Er,5Vn8P$

- Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

- (EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).

```python
# This function randomly generates a safe key with: key length provided by user, uppercase = True, lowercase = True, numbers = True, special characters = True. 
# Output copied to clipboard.

def key_generator(length, uppercase, lowercase, numbers, special_characters):  
    
    # Empty list of types of characters available
    character_types = []
    # Empty string with all the characters available
    all_characters = ''
    
    if uppercase:                                   
        character_types.append(string.ascii_uppercase)         
        all_characters += string.ascii_uppercase               
    if lowercase:                                   
        character_types.append(string.ascii_lowercase)
        all_characters += string.ascii_lowercase                
    if numbers:                                     
        character_types.append(string.digits)                  
        all_characters += string.digits                      
    if special_characters:                          
        character_types.append(string.punctuation)             
        all_characters += string.punctuation

    # Validation 
    if not all_characters:
        raise ValueError("All four types (uppercase, lowercase, numbers, special_characters) must be True. Usage: key_generator(length).")
    
    # Empty list with one character from each type
    selected_characters = []
    for each_type in character_types:
        character = np.random.choice(list(each_type))           
        selected_characters.append(character)                   

    # Calculate missing elements (if any) and add random characters until the desired length.
    remaining = length - len(selected_characters)
    if remaining > 0:
        extra_characters = np.random.choice(list(all_characters), size=remaining)     
        selected_characters.extend(extra_characters)                                  
    
    # To prevent first four character always being in the same order
    random.shuffle(selected_characters)
    
    key = ''.join(selected_characters)             
    
    pyperclip.copy(key)
    
    return key

random_key = key_generator(10, True, True, True, True)
print(random_key)  # Not required by instructions, but just to be able to see output
```
