```python
import requests
import json
```

```python
BASE_URL = "https://jsonplaceholder.typicode.com"

print("=" * 60)
print("EXPLORACIÓN DE API JSONPLACEHOLDER")
print("=" * 60)

# ==========================================================
# 1. CONSULTAS GET
# ==========================================================

print("\n1. REALIZANDO PETICIONES GET")
print("-" * 40)

response_posts = requests.get(f"{BASE_URL}/posts")
response_users = requests.get(f"{BASE_URL}/users")
response_todos = requests.get(f"{BASE_URL}/todos")

# ==========================================================
# 2. MOSTRAR CANTIDADES Y CÓDIGOS DE ESTADO
# ==========================================================

print("\n2. RESULTADOS DE LAS PETICIONES")
print("-" * 40)

posts_data = response_posts.json()
users_data = response_users.json()
todos_data = response_todos.json()

print(f"Publicaciones (/posts)")
print(f"  - Cantidad total: {len(posts_data)}")
print(f"  - Código de estado: {response_posts.status_code}")

print(f"\nUsuarios (/users)")
print(f"  - Cantidad total: {len(users_data)}")
print(f"  - Código de estado: {response_users.status_code}")

print(f"\nTareas (/todos)")
print(f"  - Cantidad total: {len(todos_data)}")
print(f"  - Código de estado: {response_todos.status_code}")

# ==========================================================
# 3. PETICIÓN A RECURSO INEXISTENTE (ERROR 404)
# ==========================================================

print("\n3. PETICIÓN A PUBLICACIÓN INEXISTENTE")
print("-" * 40)

response_not_found = requests.get(f"{BASE_URL}/posts/99999")
print(f"URL consultada: {BASE_URL}/posts/99999")
print(f"Código de estado: {response_not_found.status_code}")

if response_not_found.status_code == 404:
    print("  → Recurso no encontrado (404)")

# ==========================================================
# 4. PETICIÓN POST (CREAR NUEVA PUBLICACIÓN)
# ==========================================================

print("\n4. PETICIÓN POST - CREAR NUEVA PUBLICACIÓN")
print("-" * 40)

new_post = {
    "title": "Mi nuevo post desde Python",
    "body": "Este es el contenido del post creado mediante una petición POST a la API.",
    "userId": 1
}

response_post = requests.post(f"{BASE_URL}/posts", json=new_post)

print(f"Código de estado: {response_post.status_code}")
print("Respuesta JSON:")
print(json.dumps(response_post.json(), indent=2, ensure_ascii=False))

# ==========================================================
# 5. PETICIÓN PATCH (MODIFICAR PARCIALMENTE)
# ==========================================================

print("\n5. PETICIÓN PATCH - MODIFICAR PUBLICACIÓN EXISTENTE")
print("-" * 40)

patch_data = {
    "title": "Título modificado parcialmente"
}

response_patch = requests.patch(f"{BASE_URL}/posts/1", json=patch_data)

print(f"Código de estado: {response_patch.status_code}")
print("Respuesta JSON:")
print(json.dumps(response_patch.json(), indent=2, ensure_ascii=False))

# ==========================================================
# 6. PETICIÓN DELETE (ELIMINAR PUBLICACIÓN)
# ==========================================================

print("\n6. PETICIÓN DELETE - ELIMINAR PUBLICACIÓN")
print("-" * 40)

response_delete = requests.delete(f"{BASE_URL}/posts/1")

print(f"Código de estado: {response_delete.status_code}")
print("Respuesta JSON:")

if response_delete.status_code == 200:
    print(json.dumps(response_delete.json(), indent=2, ensure_ascii=False))
else:
    print(f"  → {response_delete.json() if response_delete.text else 'Respuesta vacía (eliminación exitosa)'}")

print("\n" + "=" * 60)
print("FIN DE LA EXPLORACIÓN")
print("=" * 60)
```
